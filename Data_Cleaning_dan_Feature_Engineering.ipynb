## 1. Mengapa Preprocessing Penting?

Preprocessing merupakan tahapan untuk menyiapkan data agar layak digunakan pada proses analisis lanjutan dan pemodelan machine learning. Permasalahan umum pada data industri meliputi nilai hilang, format yang tidak konsisten, data kategorikal yang belum dapat dibaca model, skala numerik yang berbeda jauh, dan variabel yang belum cukup representatif.

Dalam konteks teknik industri, preprocessing penting karena data operasional sering berasal dari berbagai sumber, seperti mesin, operator, inspeksi kualitas, dan sistem produksi. Oleh sebab itu, sebelum melakukan prediksi, data perlu dibersihkan dan ditransformasikan secara sistematis.

## 5. Penanganan Missing Value

Missing value dapat ditangani dengan beberapa pendekatan, seperti menghapus baris, menghapus kolom, atau melakukan imputasi. Pada sesi ini digunakan imputasi sederhana agar mahasiswa memahami prinsip dasar tanpa kompleksitas yang berlebihan.

- Variabel numerik diisi dengan nilai rata-rata.
- Variabel kategorikal dapat diisi dengan modus jika diperlukan.

```python
# =============================================================================
# MENANGANI MISSING VALUE PADA KOLOM NUMERIK
# =============================================================================
# fillna() digunakan untuk mengganti nilai kosong.
# Di sini, nilai kosong pada Downtime_Menit dan Defect_Unit
# diganti dengan rata-rata kolom masing-masing.

df['Downtime_Menit'] = df['Downtime_Menit'].fillna(df['Downtime_Menit'].mean())
df['Defect_Unit'] = df['Defect_Unit'].fillna(df['Defect_Unit'].mean())

# Memeriksa ulang missing value
df.isnull().sum()
```

Pada tahap ini, mahasiswa perlu memahami bahwa data duplikat dapat mengganggu interpretasi hasil. Pemeriksaan duplikasi merupakan prosedur standar dalam preprocessing data.

## 7. Feature Engineering: Membentuk Variabel Turunan

Feature engineering adalah proses membangun variabel baru dari data yang sudah ada agar lebih representatif terhadap fenomena yang dianalisis. Dalam konteks teknik industri, variabel turunan sering kali lebih informatif dibandingkan variabel mentah.

Pada sesi ini akan dibuat:
- `Selisih_Target` = Output aktual - Target
- `Defect_Rate` = Defect / Output
- `Produktivitas_Operator` = Output / Operator hadir

```python
# =============================================================================
# MEMBENTUK FITUR TURUNAN
# =============================================================================
# Kolom baru dibuat dengan operasi matematika sederhana antarvariabel.

df['Selisih_Target'] = df['Output_Unit'] - df['Target_Unit']
df['Defect_Rate'] = df['Defect_Unit'] / df['Output_Unit']
df['Produktivitas_Operator'] = df['Output_Unit'] / df['Operator_Hadir']

# Menampilkan hasil fitur baru
df.head()
```

## 8. Encoding Variabel Kategorikal

Sebagian besar algoritma machine learning membutuhkan input numerik. Oleh karena itu, variabel kategorikal seperti `Shift`, `Mesin`, dan `Status_Keterlambatan` perlu diubah menjadi bentuk numerik.

Pada sesi ini digunakan `LabelEncoder` untuk pengenalan awal. Pada tahap lebih lanjut, mahasiswa dapat diperkenalkan pada one-hot encoding.

```python
# =============================================================================
# ENCODING DATA KATEGORIKAL
# =============================================================================
# LabelEncoder mengubah kategori teks menjadi angka.
# Contoh: Pagi -> 0, Siang -> 1 (tergantung hasil encoder)

le_shift = LabelEncoder()
le_mesin = LabelEncoder()
le_status = LabelEncoder()

df['Shift_Encoded'] = le_shift.fit_transform(df['Shift'])
df['Mesin_Encoded'] = le_mesin.fit_transform(df['Mesin'])
df['Status_Encoded'] = le_status.fit_transform(df['Status_Keterlambatan'])

df[['Shift','Shift_Encoded','Mesin','Mesin_Encoded','Status_Keterlambatan','Status_Encoded']].head()
```

## 9. Scaling Variabel Numerik

Feature scaling diperlukan ketika variabel numerik memiliki skala yang sangat berbeda. Misalnya, `Downtime_Menit` dan `Output_Unit` memiliki rentang nilai yang berbeda. Beberapa algoritma machine learning, khususnya yang berbasis jarak, sangat sensitif terhadap perbedaan skala.

Pada sesi ini digunakan `StandardScaler`, yang mengubah data sehingga memiliki rata-rata mendekati nol dan simpangan baku mendekati satu.

```python
# =============================================================================
# STANDARDISASI FITUR NUMERIK
# =============================================================================
# StandardScaler mengubah data numerik menjadi skala standar.
# Kolom yang dipilih adalah variabel numerik utama dan fitur turunan.

scaler = StandardScaler()

numeric_cols = ['Output_Unit', 'Downtime_Menit', 'Defect_Unit', 'Suhu_Mesin',
                'Selisih_Target', 'Defect_Rate', 'Produktivitas_Operator']

scaled_data = scaler.fit_transform(df[numeric_cols])

# Membuat DataFrame baru agar hasil lebih mudah dibaca
df_scaled = pd.DataFrame(scaled_data, columns=[col + '_Scaled' for col in numeric_cols])

df_scaled.head()
```

## 10. Menyusun Dataset Siap Model

Setelah cleaning, feature engineering, encoding, dan scaling selesai, langkah berikutnya adalah menyusun dataset yang siap digunakan untuk machine learning.

```python
# Menggabungkan beberapa kolom hasil preprocessing ke dalam satu tabel akhir
df_final = pd.concat([
    df[['Shift_Encoded', 'Mesin_Encoded', 'Status_Encoded']],
    df_scaled
], axis=1)

df_final.head()
```
